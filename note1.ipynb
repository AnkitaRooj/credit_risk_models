```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data"
df = pd.read_csv(url, sep=" ", header=None)
df.columns = ['checking_account', 'duration', 'credit_history', 'purpose', 'credit_amount', 
                'savings_account', 'employment', 'installment_rate', 'personal_status', 'debtors', 
                'residence_since', 'property', 'age', 'other_installment_plans', 'housing', 
                'existing_credits', 'job', 'people_liable', 'telephone', 'foreign_worker', 'target']
df.head()
```

```python
# target variable: 1 - good, 2 - bad
df['target'].value_counts()
```

```python
df['target_cat'] = df['target'].map({1: 'good', 2: 'bad'})
```

```python
sns.countplot(x = df['target_cat'], palette = 'Set1')
plt.show()
```

```python
df['target'] = df['target'].map({1:0, 2:1})  # 0 - good, 1 - bad
df['target'].value_counts()
```
